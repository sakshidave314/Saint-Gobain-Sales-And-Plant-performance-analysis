```python
df.dropna(subset=['order_id'],inplace = True)
```

```python
df.isnull().sum()
```

```python
df["quantity_units"] = df.groupby("product_type")["quantity_units"]\
                         .transform(lambda x: x.fillna(x.median()))
```

```python
df["quantity_units"].isnull().sum()
```

```python
df["plant"] = df["plant"].fillna(df["plant"].mode()[0])

df["shift"] = df["shift"].fillna(df["shift"].mode()[0])

df["region"] = df["region"].fillna(df["region"].mode()[0])
```

```python
df["unit_cost_inr"].dtype
```

```python
df["unit_cost_inr"] = pd.to_numeric(df["unit_cost_inr"], errors="coerce")
```

```python
df["unit_cost_inr"] = df["unit_cost_inr"].fillna(df["unit_cost_inr"].median())
```

```python
df["delivery_delay_days"] = pd.to_numeric(
    df["delivery_delay_days"], errors="coerce"
)

df["delivery_delay_days"] = df["delivery_delay_days"].fillna(
    df["delivery_delay_days"].median()
)
```

```python
df["discount_pct"] = pd.to_numeric(
    df["discount_pct"], errors="coerce"
)

df["discount_pct"] = df["discount_pct"].fillna(
    df["discount_pct"].median()
)
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
df["order_date"].head(10)
```

```python
print(df.iloc[[1, 5, 7, 8]]["order_date"])
```

```python
df["product_type"] = df["product_type"].str.lower()
```

```python
df['product_type'].unique()
```

```python
df.replace({
    "product_type": {
        "glass bottles": "glass bottle",
        "temp. glass": "tempered glass",
        "igu": "insulated glass unit",
        "lam. glass": "laminated glass",
        "tempered glas": "tempered glass",
        "float glas": "float glass"
    }
}, inplace=True)
```

```python
df["product_type"] = df["product_type"].str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
for value in df["product_type"].unique():
    print(repr(value))
```

```python
df["plant"] = df["plant"].str.lower()
```

```python
df["plant"] = df["plant"].replace({
    "plant a": "howrah",
    "plant a-howrah": "howrah",
    "plant a - howrah": "howrah",
    "howrah plant": "howrah",

    "plant b": "pune",
    "plant b-pune": "pune",
    "plant b - pune": "pune",
    "pune plant": "pune",

    "plant c": "chennai",
    "plant c-chennai": "chennai",
    "plant c - chennai": "chennai",
    "chennai plant": "chennai"
})
```

```python
df["plant"].sample(20)
```

```python
df["plant"] = df["plant"].str.strip()
```

```python
df["region"] = df["region"].str.lower()
```

```python
df["region"] = df["region"].replace({
    "n": "north",
    "nort": "north",
   

    "south ": "south",
    "s": "south",
    

    "eest": "east",
    "wast": "west",
    "centeral": "central"
})
```

```python
for value in df["region"].unique():
    print(repr(value))
```

```python
df["unit_price_inr"] = pd.to_numeric(
    df["unit_price_inr"],
    errors="coerce"
)
```

```python
df["unit_price_inr"].dtype
```

```python
float_cols = df.select_dtypes(include="float64").columns
df[float_cols] = df[float_cols].round(2)
```

```python
df["customer_complaint"] = df["customer_complaint"].replace({
    "True": "Yes",
    "1": "Yes",
    "False": "No",
    "0": "No"
})
```

```python
df["customer_complaint"] = df["customer_complaint"].fillna("No")
```
